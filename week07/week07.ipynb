### Set Operators ###


```python
# UNIONS #
A = {1,2,3,5,7} 
B = {5,7,9,13,15,18} 
print(A.union(B)) 
print(B.union(A)) # union = A + B both values
```

```python
A = {1,2,3,5,7}
B = {5,7,9,13,15,18}
print(A|B)
print(B|A)   
```

```python
A = {1,2,3,5,7}
B = {5,7,9,13,15,18}
print(A|B)
```

```python
# INTERSECTION
A = {1,2,3,5,7}
B = {5,7,9,13,15,18}
print(A & B)
print(B & A)   it gives commen values 
```

```python
A = {1,2,3,5,7}
B = {5,7,9,13,15,18}
print(A.intersection(B))
print(B.intersection(A))
```

```python
# Difference
A = {1,2,3,5,7}
B = {5,7,9,13,15,18}
print(A.difference(B))
print(B.difference(A))
```

```python
student = {"name":"nusaiba","age":23,"city":"batticaloa"}
print(student["name"])
```
